# Figure notebook 04 — TinyViT on MNIST even/odd

Paper figures, built from the bundles `notebooks/04_ViT.ipynb` exports — no model, no
dataset, no BFT run. Everything here works on any checkout of the repo.

- data: `figures/figdata/{nb04_circuits, nb04_fingerprints}.npz`
- render functions: `src/paper_figures.py` (`FIGURES` registry)
- style/venue: `figstyle.py`, config in `.figstyle/`, captions in `.figstyle/figures.md`
- regenerate every figure of the paper at once: `python scripts/render_figures.py`

## §2 — What is in the bundle

`summary()` prints every key with its shape and dtype — enough to design a panel
without opening the analysis notebook. The trace itself is under `nodes` (one
entry per BFT node, breadth-first: lambdas, per-stimulus loadings, connection
maps, class profiles, top stimuli, weighted-average inputs), with a few real
stimuli under `stimuli`.

```python
figdata.summary('nb04_circuits')
figdata.summary('nb04_fingerprints')
```
