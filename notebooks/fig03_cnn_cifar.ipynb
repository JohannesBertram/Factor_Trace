# Figure notebook 03 — SmallCNN on CIFAR-10

dataset, no BFT run. Everything here works on any checkout of the repo.

- render functions: `src/paper_figures.py` (`FIGURES` registry)
- style/venue: `figstyle.py`, config in `.figstyle/`, captions in `.figstyle/figures.md`
- regenerate every figure of the paper at once: `python scripts/render_figures.py`

## §2 — What is in the bundle

without opening the analysis notebook. The trace itself is under `nodes` (one

stimuli under `stimuli`.
